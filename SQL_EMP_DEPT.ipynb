## Оконные функции

Простой пример

```python
sql = '''select e.ename,
e.deptno,
count(*) over () as cnt
from EMP e
order by 2'''
```

```python
select(sql)
```

Порядок обработки

```python
sql = '''select e.ename,
e.deptno,
count(*) over () as cnt
from EMP e
where deptno = 10
order by 2'''
```

```python
select(sql)
```

Сегменты или группа строк

```python
sql = '''select e.ename,
e.deptno,
count(*) over(partition by e.deptno) as cnt
from EMP e
order by 2 '''
```

```python
select(sql)
```

Предшествующий запрос можно рассматривать как более эффективную версию следующего запроса

```python
sql = '''select e.ename,
e.deptno,
(select count(*) from emp d
where e.deptno=d.deptno) as cnt
from emp e
order by 2'''
```

```python
select(sql)
```

Еще одно полезное свойство оператора PARTITION BY состоит в том что он выполняет вычисления независимо от других оконных функций осуществляя сегментирование по другим столбцам в том же выражении SELECT

```python
sql = '''select e.ename,
e.deptno,
count(*) over (partition by e.deptno) as dept_cnt,
job,
count(*) over (partition by e.job) as job_cnt
from emp e
order by 2 '''
```

```python
select(sql)
```

Если в функцию COUNT вместо * передать COMM то результаты будут значительно иными. При использовании COUNT(*) считатются не значения столбцов а строки т.е. NULL учитывается

Когда порядок имеет значение

```python
sql = '''select
e.deptno,
e.ename,
e.hiredate,
e.sal,
sum(sal) over(partition by e.deptno) as total1,
sum(sal) over() as total2,
sum(sal) over(order by hiredate) as running_total
from emp e
where deptno =10'''
```

```python
select(sql)
```

Следующий запрос такой же как и предыдущий но в нем используется оператор RANGE BETWEEN чтобы явно задать поведение по умолчанию в результате применения ORDER BY HIREDATE

```python
sql = '''select
e.deptno,
e.ename,
e.hiredate,
e.sal,
sum(sal) over(partition by e.deptno) as total1,
sum(sal) over() as total2,
sum(sal) over(order by hiredate
                  range between unbounded preceding
                  and current row) as running_total
from emp e
where deptno =10'''
```

```python
select(sql)
```

Оператор кадрирования (RANGE BETWEEN)

```python
sql = ''' select
e.deptno,
e.ename,
e.sal,
sum(sal) over(order by e.hiredate
              range between unbounded preceding
              and current row) as run_total1,
sum(sal) over(order by e.hiredate
              rows between 1 preceding
              and current row) as run_total2,
sum(sal) over(order by e.hiredate
              range between current row
              and unbounded following) as run_total3,
sum(sal) over(order by e.hiredate
              rows between current row
              and 1 following) as run_total4
from emp e
where deptno = 10'''
```

```python
select(sql)
```

```python
sql = '''select
ename,
sal,
min(sal) over(order by sal) min1,
max(sal) over(order by sal) max1,
min(sal) over(order by sal
              range between unbounded preceding
              and  unbounded following) min2,
max(sal) over(order by sal
              range between unbounded preceding
              and unbounded following) max2,
min(sal) over(order by sal
              range between current row
              and current row) min3,
max(sal) over(order by sal
              range between current row
              and current row) max3,
max(sal) over(order by sal
              rows between 3 preceding
              and 3 following) max4
from emp
'''
```

```python
select(sql)
```

Удобочитаемость + производительность = мощность

- количество служащих в каждом отделе;
- количество разных должностей в каждом отделе (например, количество клерков в отделе 10);
- общее количество служащих в таблице EMP

```python
sql = '''select
e.deptno,
e.job,
count(*) over(partition by e.deptno) as emp_cnt,
count(job) over(partition by e.deptno, e.job) as job_cnt,
count(*) over() as total
from emp e
'''
```

```python
select(sql)
```

чтобы получить такую же информацию без использования оконных функций потребовалось бы выполнить несколько больший обьем работ

```python
sql = '''select a.deptno, a.job,
(select count(*) from emp b
where b.deptno = a.deptno) as emp_cnt,
(select count(*) from emp b
where b.deptno = a.deptno and b.job = a.job) as job_cnt,
(select count(*) from emp) as total
from emp a
order by 1,2
 '''
```

```python
select(sql)
```

## Рекурсивный подзапрос

Считаю до пяти

```python
sql = '''with recursive counter(n) as(
      select 1
      union all
      select n+1
      from counter
      where n<5
    )
    select n from counter '''
```

```python
select(sql)
```

Добавил индекс (в даном случае индекс - счетчик)

```python
sql = '''with recursive counter(n, index1) as(
      select 1,0
      union all
      select n+1, index1+1
      from counter
      where index1<5
    )
    select n, index1 from counter '''
```

```python
select(sql)
```

Вычисление первых 20 значений последовательности Фибоначчи

```python
sql = '''with recursive workingTable(fibNum, NextNumber, index1)
as
(select 0,1,1
union all
select fibNum+nextNumber, fibNUm, index1+1
from workingTable
where index1<20)
select fibNum from workingTable as fib '''
```
