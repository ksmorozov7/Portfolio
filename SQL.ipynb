Это запрос строит декартово произведение (все возможные пары).
Как он работает по шагам
1. CTE users — создаёт временный набор из трёх строк: user_id = 1, 2, 3.
2. CTE month — создаёт временный набор из двух строк: month = '2021-03-01' и '2021-04-01'.
3. JOIN ... ON 1=1 — это хитрый приём: условие 1=1 всегда истинно, поэтому каждая строка из users соединяется с каждой строкой из month.

Зачем так делают
1. Создать «сетку» для отчётов. Например, чтобы для каждого пользователя получить строку за каждый месяц — даже если в реальных данных за какой‑то месяц записей нет. Потом к этой сетке можно присоединить фактические данные через LEFT JOIN и увидеть пропуски.
2. Подготовить шаблон для агрегаций. Удобно, когда нужно явно иметь все комбинации категорий, чтобы корректно посчитать метрики по всем срезам.
3. Тестирование логики. Как и в прошлом примере, это быстрый способ получить небольшой детерминированный набор данных для проверки запроса.

Важное уточнение
1. Это не создаёт таблиц в базе — оба CTE (users и month) существуют только в рамках выполнения этого запроса.

```python
sql = '''
with users as(
select 1 as user_id
union all
select 2 as user_id
union all
select 3 as user_id),

mont as

(select date('2021-03-01') as month
union all
select date ('2021-04-01') as month)

select * from users t
join month m on 1=1

'''
```

```python
select(sql)
```

```python
sql = f'''WITH dates(month) AS (
  VALUES(({min}))
  UNION ALL
  SELECT date(month, '+1 month')
  FROM dates
  WHERE month < ({max})
),

clients as (select distinct t.client_id from german_credit t),

clients_month as

(SELECT t.month, c.client_id FROM dates t
join clients c on 1=1),

trans_month as

(select
date(t.dt, 'start of month') as month,
t.client_id,
count(1) as transaction_cnt,
sum(t.amount) as amount_sum

 from client_transactions t
group by 1,2
order by 1),

client_trans_month as (

select t.*,
tm.transaction_cnt,
tm.amount_sum,
1 as user,
case when tm.transaction_cnt > 0 then 1 else 0 end as active

 from clients_month t
left join trans_month tm on t.client_id = tm.client_id
and t.month = tm.month
)

select t.month, sum(t.user) as user_cnt, sum(t.amount_sum) as amount_sum ,
sum(t.active) as active_cnt from client_trans_month t
group by t.month
'''
```
